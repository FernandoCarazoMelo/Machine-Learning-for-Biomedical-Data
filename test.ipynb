```python
import numpy as np
import pandas as pd

np.random.seed(42)
n_samples = 1000

age = np.random.randint(1, 20, n_samples)
mileage = age * np.random.randint(8000, 15000, n_samples)
horsepower = np.random.randint(70, 400, n_samples)
brand_tier = np.random.choice([1, 2, 3], n_samples)  # 1=budget, 2=mid, 3=luxury

price = (
    30000
    - age * 1500
    - mileage * 0.05
    + horsepower * 50
    + brand_tier * 5000
    + np.random.normal(0, 3000, n_samples)
).clip(1000)

df = pd.DataFrame({
    'age': age,
    'mileage': mileage,
    'horsepower': horsepower,
    'brand_tier': brand_tier,
    'price': price
})

print(df.shape)
df.head()
```
